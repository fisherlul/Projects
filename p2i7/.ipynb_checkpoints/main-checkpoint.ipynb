```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.animation import FuncAnimation
import numpy as np
import sympy as sp
%matplotlib widget
```

```python
def lennard_jones(r, eps=1, sigma=1, equilibrium=False):
    r_cutoff = 2.5 * sigma
    if r == 0 or r > r_cutoff:
        return 0
    if equilibrium:
        r = 1.122 * sigma
    return 4 * eps * ((sigma/r)**12 - (sigma/r)**6)

def force(r):
    V = lennard_jones(r)
    return sp.diff(V, r)

def create_square_lattice(n, spacing):
    atoms = []
    for i in range(n):
        for j in range(n):
            atoms.append((i * spacing, j * spacing))
    return np.array(atoms)

def create_hexagonal_lattice(n, spacing):
    atoms = []
    for i in range(n):
        for j in range(n):
            x = i * spacing + (j % 2) * spacing / 2       # spacing if same even row, else /2 if odd
            y = j * spacing * np.sqrt(3) / 2              # vertical spacing based on geometry (* sin(60))
            atoms.append((x, y))
    return np.array(atoms)
```

```python
def compute_total_energy(atoms):
    energy = 0
    for i in range(len(atoms)):
        for j in range(i + 1, len(atoms)):
            r = np.linalg.norm(atoms[i] - atoms[j])
            energy += lennard_jones(r)
    return energy

def compute_local_energy(atoms):
    energies = np.zeros(len(atoms))
    for i in range(len(atoms)):
        for j in range(len(atoms)):
            if i != j:
                r = np.linalg.norm(atoms[i] - atoms[j])
                energies[i] += lennard_jones(r)
    return energies
```

```python
# Parameters
n = 10
spacing = 0.8      # distance entre atomes

# Analyse énergétique en fonction du spacing
spacings = np.linspace(n/10, 1.5)
square_energies = []
hex_energies = []

for s in spacings:
    square_atoms = create_square_lattice(n, s)
    hex_atoms = create_hexagonal_lattice(n, s)
    square_energies.append(compute_total_energy(square_atoms))
    hex_energies.append(compute_total_energy(hex_atoms))
    
# Énergies
square_energy = compute_total_energy(square_atoms)
hex_energy = compute_total_energy(hex_atoms)

square_local = compute_local_energy(square_atoms)
hex_local = compute_local_energy(hex_atoms) 
    
square_min_spacing = spacings[np.argmin(square_energies)]
hex_min_spacing = spacings[np.argmin(hex_energies)]
print(square_min_spacing)
print(hex_min_spacing)

# Tracé de l'énergie en fonction du spacing
plt.figure(figsize=(10, 6))
plt.plot(spacings, square_energies, label="Carré", color="blue")
plt.plot(spacings, hex_energies, label="Hexagonal", color="green")
plt.axvline(square_min_spacing, color="red", linestyle="--", label="Min Carré")
plt.axvline(hex_min_spacing, color="orange", linestyle="--", label="Min Hexagonal")
plt.xlabel("Distance inter-atomique (spacing)")
plt.ylabel("Énergie totale du réseau")
plt.title("Énergie de Lennard-Jones en fonction d'intéraction inter-atomique")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
def draw_bonds(ax, atoms, threshold=1.3):
    for i in range(len(atoms)):
        for j in range(i + 1, len(atoms)):
            r = np.linalg.norm(atoms[i] - atoms[j])
            if r < threshold:
                ax.plot(*zip(atoms[i], atoms[j]), color='white', linewidth=0.5)

def plot_lattice_with_bonds(ax, atoms, energies, title):
    norm = plt.Normalize(vmin=min(energies), vmax=max(energies))
    draw_bonds(ax, atoms, threshold=1.3)  # Add bonding lines
    for (x, y), e in zip(atoms, energies):
        circle = Circle((x, y), 0.2, color=plt.cm.viridis(norm(e)))
        ax.add_patch(circle)
    ax.set_aspect('equal')
    ax.set_xlim(min(atoms[:, 0]) - 1, max(atoms[:, 0]) + 1)
    ax.set_ylim(min(atoms[:, 1]) - 1, max(atoms[:, 1]) + 1)
    ax.axis('off')
    ax.set_title(title)

# Visualisation des réseaux
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
plot_lattice_with_bonds(ax1, square_atoms, square_local, f"Carré\nÉnergie totale: {square_energy:.2f}")
plot_lattice_with_bonds(ax2, hex_atoms, hex_local, f"Hexagonal\nÉnergie totale: {hex_energy:.2f}")
plt.suptitle("Comparaison de réseaux 2D avec le potentiel de Lennard-Jones")
plt.tight_layout()
plt.show()
```

```python
# Parameters
box_size = 10  # size of the square box
n_particles = 3
n_steps = 1000  # number of time steps
dt = 0.1       # time step size

# Initial positions and velocities of particles
positions = np.array([
    [2.0, 3.0],
    [5.0, 5.0],
    [8.0, 1.0]
])

velocities = np.array([
    [0.9, -0.4],
    [-0.6, 0.8],
    [0.2, 1.2]
])

# Time array
time = np.arange(0, n_steps * dt, dt)

# Store position history
positions_history = np.zeros((n_steps, n_particles, 2))
positions_history[0] = positions

# Simulation loop
for i in range(1, n_steps):
    # Update positions
    positions += velocities * dt
    
    # Apply periodic boundary conditions (wrap around)
    positions = positions % box_size

    # Store positions
    positions_history[i] = positions

fig, ax = plt.subplots()
sc = ax.scatter([], [], s=100)
ax.set_xlim(0, box_size)
ax.set_ylim(0, box_size)
ax.set_aspect('equal')
ax.set_title("Particle Movement in a 10x10 Box")

def update(frame):
    sc.set_offsets(positions_history[frame])
    return sc,

ani = FuncAnimation(fig, update, frames=n_steps, interval=50, blit=True)
plt.show()
```

```python
# Parameters
num_steps = 500       # Number of steps in the simulation
step_size = 0.5       # How far the particle moves each step

# Initialize position arrays
x = np.zeros(num_steps)
y = np.zeros(num_steps)

# Generate Brownian motion
for i in range(1, num_steps):
    angle = np.random.uniform(0, 2 * np.pi)  # Random direction
    x[i] = x[i-1] + step_size * np.cos(angle)
    y[i] = y[i-1] + step_size * np.sin(angle)

# Animation function
fig, ax = plt.subplots()
line, = ax.plot([], [], lw=2)
dot, = ax.plot([], [], 'ro')  # Current position of the particle

ax.set_xlim(np.min(x)-1, np.max(x)+1)
ax.set_ylim(np.min(y)-1, np.max(y)+1)
ax.set_title("2D Brownian Motion of a Particle")

def init():
    line.set_data([], [])
    dot.set_data([], [])
    return line, dot

def update(frame):
    line.set_data(x[:frame], y[:frame])
    dot.set_data(x[frame-1], y[frame-1])
    return line, dot

ani = FuncAnimation(fig, update, frames=num_steps, init_func=init, blit=True, interval=30, repeat=False)

plt.show()
```

```python
# Animation function
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

# Plot for particle motion
line, = ax1.plot([], [], lw=2)
dot, = ax1.plot([], [], 'ro')  # Current position of the particle
ax1.set_xlim(np.min(x)-1, np.max(x)+1)
ax1.set_ylim(np.min(y)-1, np.max(y)+1)
ax1.set_title("2D Particle Crystallization")

# Plot for speed tracking
speed_line, = ax2.plot([], [], lw=2, label="Speed")
ax2.set_xlim(0, len(x))
ax2.set_ylim(0, 1.1 * np.max(step_size))
ax2.set_title("Particle Speed Over Time")
ax2.set_xlabel("Step")
ax2.set_ylabel("Speed")
ax2.legend()

def init():
    line.set_data([], [])
    dot.set_data([], [])
    speed_line.set_data([], [])
    return line, dot, speed_line

def update(frame):
    line.set_data(x[:frame], y[:frame])
    dot.set_data(x[frame-1], y[frame-1])
    
    if frame > 1:
        speeds = np.sqrt(np.diff(x[:frame])**2 + np.diff(y[:frame])**2)
        speed_line.set_data(range(1, frame), speeds)
    else:
        speed_line.set_data([], [])
        
    return line, dot, speed_line


ani = FuncAnimation(fig, update, frames=len(x), init_func=init, blit=True, interval=30, repeat=False)

plt.tight_layout()
plt.show()
```

```python
# Thiết lập thông số
num_particles = 4  # Số phân tử
num_steps = 200    # Số bước chuyển động
step_size = 0.05   # Kích thước bước di chuyển (càng nhỏ, chuyển động càng mượt)

# Tạo vị trí ban đầu ngẫu nhiên cho các phân tử
np.random.seed(42)  # Để kết quả có thể tái hiện
positions = np.random.rand(num_particles, 2) * 10  # Mảng 2D: [x, y]

# Thiết lập figure
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_title("Mô phỏng Chuyển động Brown của 4 Phân Tử trong Chất Lỏng")

# Màu sắc và kích thước phân tử
colors = ['red', 'blue', 'green', 'purple']
particles = [ax.plot([], [], 'o', markersize=10, color=colors[i])[0] 
             for i in range(num_particles)]

# Hàm cập nhật vị trí các phân tử
def update():
    global positions
    # Thêm nhiễu ngẫu nhiên vào vị trí (mô phỏng va chạm phân tử)
    positions += np.random.normal(0, step_size, (num_particles, 2))
    
    # Đảm bảo phân tử không đi ra khỏi khung
    positions = np.clip(positions, 0, 10)
    
    # Cập nhật vị trí mới
    for i, particle in enumerate(particles):
        particle.set_data(positions[i, 0], positions[i, 1])
    
    return particles

# Tạo animation
ani = FuncAnimation(fig, update, frames=num_steps, interval=50, blit=True)
plt.show()
```
